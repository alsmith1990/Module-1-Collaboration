```python
# Name: Andrew Smith
# App Name: CGNU GPA Checker (the CGNU name is just for fun, don't worry about it)
# This app asks for a student's last and first name, then their GPA. It then tellss the user if the GPA the input says it is qualified for 
# Honor Roll, Dean's List or neither.

while True:
    last_name = input("Enter student's last name (or 'ZZZ' to quit): ")
    if last_name == 'ZZZ':
        break
        
    first_name = input("Enter student's first name: ")
    gpa = float(input("Enter student's GPA: "))
    
    if gpa >= 3.5:
        print(f"{first_name} {last_name} has made the Dean's List.")
    elif gpa >= 3.25:
        print(f"{first_name} {last_name} has made the Honor Roll.")
    else:
        print(f"{first_name} {last_name} does not qualify for either.")
```
